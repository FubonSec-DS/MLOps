
# 🧙‍♂️ vdclient Magic SDK 使用教學
vdclient magic 除了可在 Jupyter Notebook 環境中使用，也可以透過 SDK 在 .py 程式內執行

```python
import pandas as pd
import os
from vdclient_magic.core.sql import sql_to_df, sql_to_file
```

## 🪄 sql_to_df：將 SQL 語法回傳結果存成 python dataframe，用來取用資料進行後續資料分析

```python
sql = "SELECT * FROM provider_data"
df = sql_to_df(sql, datasource="insurance", chunksize=200000) # insurance 為在 DeepFlow 資料源連線管理上設定的「資料源名稱」

print(f"✅ 合併後資料列數：{df.shape[0]}")
display(df.head())
```

## 🪄 sql_to_file：將 SQL 語法回傳結果存成 parquet 格式檔案，用於大量資料存取

```python
# 暫存 Parquet 檔案
filename = "/tmp/test_sql_output.parquet"
sql = "SELECT * FROM provider_data"

# 寫入
sql_to_file(sql, filename, datasource="insurance", chunksize=200000)

# 驗證讀回內容
if os.path.exists(filename):
    df = pd.read_parquet(filename)
    print(f"✅ 成功寫入 parquet，列數：{df.shape[0]}")
    display(df.head())
    os.remove(filename)
else:
    print("❌ 檔案寫入失敗")
```
