
# 🧙‍♂️ vdclient Magic 使用教學

這份 Notebook 是為了協助 **第一次使用 vdclient magic 的使用者** 能夠快速上手。
vdclient magic 是一套支援多種 RDB 資料來源的查詢工具，特別設計給在 Jupyter Notebook 環境中進行互動式開發的資料科學家與分析師使用。


### %vdclient -h：說明vdclient使用方式

```python
%vdclient -h
```

### 🪄[常見用法一] %%vdclient (grid): 顯示 SQL 語法前一千筆回傳結果，可用來確認 SQL 語法是否如預期 

```python
## 備註："insurance" 為在 DeepFlow 資料源連線管理上設定的「資料源名稱」 
```

```python
%%vdclient --datasource insurance
SELECT * FROM provider_data
```

### 🪄[常見用法二] %%vdclient -v variable: 將 SQL 語法回傳結果存成 python dataframe，用來取用資料進行後續資料分析

```python
%%vdclient -v data --datasource insurance
SELECT * FROM provider_data
```

### 🪄[常見用法三] %%vdclient -e file.parquet: 將 SQL 語法回傳結果存成 parquet 格式檔案，用於大量資料存取

```python
%%vdclient -e "/tmp/test_postgres.parquet" --datasource insurance
SELECT * FROM provider_data
```

```python
import pandas as pd

df = pd.read_parquet("/tmp/test_postgres.parquet")

print('shape:{}'.format(df.shape))
df.head()
```

### 🪄[常見用法四] 支援動態參數傳遞
